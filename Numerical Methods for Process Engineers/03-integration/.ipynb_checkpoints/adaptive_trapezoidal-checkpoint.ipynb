# Adaptive Trapezoidal Rule

The trapezoidal method is a fast and accurate method, if appropriate segments are chosen. In this function we will double the amount of segments until a desired relative residual is met.

## 1 — Imports

```python
import numpy as np
import matplotlib.pyplot as plt

# Using the plotting standards from the reference notebook (even if unused here)
plt.rcParams.update({
    'figure.dpi': 120,
    'axes.grid': True,
    'grid.alpha': 0.3,
    'font.size': 11
})
```

## 2 — Implementation

```python
def f(x):
    """
    The mathematical function to integrate.
    
    Parameters
    ----------
    x      : float or array-like – evaluation point(s)
    
    Returns
    -------
    result : float or array-like – sin(x)
    """
    return np.sin(x)


def trapezoidal_iter(x_lower, x_upper, relative_residual, initial_segments=2):
    """
    Trapezoidal integration method with adaptive refinement to meet a residual limit.
    
    Parameters
    ----------
    x_lower           : float – lower integration bound
    x_upper           : float – upper integration bound
    relative_residual : float – desired relative tolerance limit
    initial_segments  : int   – initial number of segments to start the iteration
    
    Returns
    -------
    I_new             : float – approximated integral value
    n_steps           : int   – number of refinement steps taken
    n_segs            : int   – final number of segments/trapezoids used
    """
    
    # calculate previous baseline value (using 1 midpoint segment)
    I_new   = (x_upper - x_lower) * f(0.5 * (x_upper + x_lower))
    
    # set initial numbers
    n_segs  = initial_segments
    n_steps = 0
    
    # loop until relative residual is met
    while True:
        I_old = I_new
        I_new = 0
        
        for i in range(0, n_segs):
            # calculate lower and upper limit of i-th segment:
            x_lower_i = x_lower + (x_upper - x_lower) / n_segs * (i)
            x_upper_i = x_lower + (x_upper - x_lower) / n_segs * (i + 1)
            
            # calculate area of i-th trapezoid
            I_new     = I_new + (x_upper_i - x_lower_i) * (0.5 * (f(x_upper_i) + f(x_lower_i)))
            
        # calculate relative residual
        if abs((I_old - I_new) / I_old) < relative_residual:
            break
            
        # increase step counter
        n_steps = n_steps + 1
        
        # double number of segments
        n_segs  = n_segs * 2
        
    return I_new, n_steps, n_segs
```

## 3 — Execution

```python
# Executing the function with a custom initial segment count of 10
integral, steps, segments = trapezoidal_iter(0, np.pi, 1e-6, initial_segments=10)

print(f"Integral : {integral}")
print(f"Steps    : {steps}")
print(f"Segments : {segments}")
```
